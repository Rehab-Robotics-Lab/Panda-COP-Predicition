```python
import scipy
import seaborn as sns
from scipy import stats
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from Analyze_models import analyze_models


cols=["ID",	"Model",	"MAE X",	"MAE Y",	"Pears X",	"Pears Y",	"Spear X",	"Spear Y",	"Mean MAE",	"Mean Pears",	"Mean Spear"]
exclude=["shift","pose","t_offset","adjust","s_offset"]
exclude=[]
include=True

# model_names=["Bi_LSTM_1L","Bi_LSTM_2L","Bi_LSTM_3L","ST_GCN_3D","ST_GCN_2D","ST_GCN_param_est"]
folder=r"C:\Users\franc\Box\Rehab Robotics Lab\Projects\PANDA Gym (# 834084)\Personnel\Students & RAs\Francis Sowande\Aim 3\Final Models\Model Data"
pro=analyze_models(folder,"val",exclude_vars=exclude,show_included=include)

metric_cols=["MAE X",	"MAE Y",	"Pears X",	"Pears Y",	"Spear X",	"Spear Y",	"Mean MAE",	"Mean Pears",	"Mean Spear"]
models=["Bi_LSTM_1L","Bi_LSTM_2L","Bi_LSTM_3L","ST_GCN_3D","ST_GCN_2D","ST_GCN_param_est","phys_best","phys_face","phys_ears"]
metrics=["Mean MAE"]
metrics=["MAE X",	"MAE Y",	"Mean MAE"]

df_list=[]
outlier_list=pd.DataFrame()
outlier_track=pd.DataFrame(columns=cols)



for model in models:
    #Loading model object
    model_object=pro.get_model_object(model)
    results=model_object.table
    print(len(results))
    print("----Model: ", model)

    #Making table similar to results for outlier tracking
    tracking_tb=results.copy()
    #Noting down model
    tracking_tb.loc[:,"Model"]=model
    #Zeroing all value to be modeified later
    tracking_tb.loc[:,metric_cols]=0 
    #Adding col with combined id, month
    tracking_tb["combined_id"]=tracking_tb["ID"].astype(int).astype(str) + "_" +tracking_tb["Month"].astype(int).astype(str)   

    for metric in metrics:
        print('--Metric: ',metric)
        #Table with outliers and their metrics
        outliers=model_object.metric_outliers(metric)
        print(len(outliers)," outliers found")
        #cominginf id_month to make unqiue identifier
        id_month=outliers["ID"].astype(int).astype(str) + "_" +outliers["Month"].astype(int).astype(str)

        tracking_tb.loc[tracking_tb["combined_id"].isin(id_month),metric]=1

        #Table with list of outliers
        out_tbl=pd.DataFrame({
            "combined_id" : id_month,
            "ID":outliers["ID"].astype(int),
            "Month":outliers["Month"].astype(int),
            "Model":model})
        outlier_list=pd.concat([outlier_list,out_tbl])

        print(id_month)

        #Table collecting metric results
        tb = pd.DataFrame({
            "Model": model,
            "metric": metric,
            "score": results[metric]  # This expands automatically if results[metric] is a Series/array
        })
        df_list.append(tb)

    outlier_track=pd.concat([outlier_track,tracking_tb])
    # print(outlier_track)

# outlier_track.to_excel(r"C:\Users\franc\Box\Rehab Robotics Lab\Projects\PANDA Gym (# 834084)\Personnel\Students & RAs\Francis Sowande\Aim 3\Final Models\outlier_tracking.xlsx")

print(outlier_list)
print(outlier_list.combined_id.value_counts(dropna=False))
print(outlier_list.Model.value_counts(dropna=False))

# Concatenate all at once
df = pd.concat(df_list, ignore_index=True)
    
color_p="gist_earth"

print(df.groupby('Model')["score"].describe())
# print(per_id["std"].describe))

plt.figure(figsize=(12,6))
sns.boxplot(df,x="Model",y="score",hue="metric",palette=color_p )
plt.show()


plt.figure(figsize=(12,6))
sns.violinplot(df,x="Model",y="score",hue="metric",palette=color_p)
plt.show()

# #Loading test results file and chaging model name from defualt "Data" to model name
# for metric in metrics:
#     plt.figure(figsize=(12,6))
#     # sns.histplot(test_data,x="Mean MAE",hue="Model",kde=True)
#     sns.kdeplot(test_data,x=metric,hue="Model")
#     # plt.legend()
#     plt.show()

#     print(test_data.groupby('Model')[metric].describe())





```
