Block to update synch times

```python
def update_synch(baby,table, vidnum,vidnames):
    col_names=[["cam1_start","cam1_stop"],["cam2_start","cam2_stop"],["cam3_start","cam3_stop"],["cam4_start","cam4_stop"],["cam5_start","cam5_stop"],["cam6_start","cam6_stop"],["cam7_start","cam7_stop"]]

    for i in [1,2,3,4,5,7]:
        st=pd.isnull(table.loc[table["vid"]==vidnum,col_names[i-1][0]]).item()
        sp=pd.isnull(table.loc[table["vid"]==vidnum,col_names[i-1][1]]).item()
        # print(st,sp)
        cam_num=i
        print("CAM: ", cam_num)
        video=vidnames[i-1]
        print(video)
        if (st and sp) :
        
            v_start_stop, v_success = baby.start_stop(video,display=0)
            # if initial video check is inaccurate,rotate video, it can make a difference
            
            table.loc[table["vid"]==vidnum,col_names[i-1][0]]=v_start_stop[0]
            table.loc[table["vid"]==vidnum,col_names[i-1][1]]=v_start_stop[1]

    # table.loc[table["vid"]==vid]=row
    return table



df= pd.read_excel(r"C:\Users\franc\Documents\Infant_Sim_data\passive sim\synchlight_time.xlsx")
# dates=pd.to_datetime(df.Date)
name="sim_trunk"

vids=[3,4,5]

sim = c_manage(cam_direct=r"C:\Users\franc\Documents\Infant_Sim_data\passive sim\Cameras", vid_name=name)

for vid_num in vids:
    print("VID:", vid_num)
    _,vid_names=sim.check_vids(vid_num)
    df=update_synch(sim, df, vid_num,vid_names)

    df.to_excel(r"C:\Users\franc\Documents\Infant_Sim_data\passive sim\synchlight_time.xlsx")  
```
