```python
def update_synch(baby,table, vid):
    col_names=[["cam1_start","cam1_stop"],["cam2_start","cam2_stop"],["cam3_start","cam3_stop"],["cam4_start","cam4_stop"],["cam5_start","cam5_stop"],["cam6_start","cam6_stop"],["cam7_start","cam7_stop"]]

    for i in range(7):
        st=pd.isnull(table.loc[table["vid"]==vid,col_names[i][0]]).item()
        sp=pd.isnull(table.loc[table["vid"]==vid,col_names[i][1]]).item()
        # print(st,sp)

        cam_num=i+1
        print("CAM: ", cam_num)
        if (st and sp) :
        
            v_start_stop, v_success = baby.start_stop(cam_num, vid,display=0)
            # if initial video check is inaccurate,rotate video, it can make a difference
            
            table.loc[table["vid"]==vid,col_names[i][0]]=v_start_stop[0]
            table.loc[table["vid"]==vid,col_names[i][1]]=v_start_stop[1]

    # table.loc[table["vid"]==vid]=row
    return table


def fix_cam3(baby,table, vid):
    col_names=[["cam1_start","cam1_stop"],["cam2_start","cam2_stop"],["cam3_start","cam3_stop"],["cam4_start","cam4_stop"],["cam5_start","cam5_stop"],["cam6_start","cam6_stop"],["cam7_start","cam7_stop"]]

    i=2
    
    st=(table.loc[table["vid"]==vid,col_names[i][0]]).item()==5
    sp=(table.loc[table["vid"]==vid,col_names[i][1]]).item()==5
    # print(st,sp)

    cam_num=i+1
    print("CAM: ", cam_num)
    if (st and sp) :
    
        v_start_stop, v_success = baby.start_stop(cam_num, vid,display=0)
        # if initial video check is inaccurate,rotate video, it can make a difference
        
        table.loc[table["vid"]==vid,col_names[i][0]]=v_start_stop[0]
        table.loc[table["vid"]==vid,col_names[i][1]]=v_start_stop[1]

    # table.loc[table["vid"]==vid]=row
    return table


userdirect = r"C:\Users\franc\Box\Rehab Robotics Lab\Projects\PANDA Gym (# 834084)\Trials"

aim=3

maindirect=userdirect + "\\Aim " + "I" * aim
df= pd.read_excel(maindirect+"\\Aim "+str(aim)+" synch light.xlsx")


# dates=pd.to_datetime(df.Date)
subid=df.subjectID
subid_u=pd.unique(subid)

tot=len(subid_u)
check=[202,211,254]

# for i in range(tot):
for id in check:
    # id=subid_u[i]
    print("SUBJECT ID: ", str(id).zfill(3))
    subject=df.loc[df['subjectID']==id]

    sub_dates_ind=(pd.unique(subject.Date))
    
    sub_dates=pd.to_datetime(sub_dates_ind)

    # print(subject)

    for j in range(len(sub_dates)):
        trial_date=sub_dates[j]
        print("Date: ", str(trial_date))
        baby = c_manage(userdirect, aim, id, trial_date.month, trial_date.day, trial_date.year)

        trial=df.loc[(df['subjectID']==id)&(df['Date']==sub_dates_ind[j])]
        # display(trial)

        vids=[4,5,6]

        for vid_num in vids:
            print("VID:", vid_num)
            trial=update_synch(baby,trial, vid_num)
            # trial=fix_cam3(baby,trial, vid_num)
            
        
        # display(trial)

        df.loc[(df['subjectID']==id)&(df['Date']==sub_dates_ind[j])]=trial
        # display(df.loc[(df['subjectID']==id)&(df['Date']==sub_dates_ind[j])])

        df.to_excel(maindirect+"\\Aim "+str(aim)+" synch light.xlsx", index=False)  
```
